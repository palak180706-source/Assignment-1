#### Conditional Statement

```python
"""
1. Electricity Bill Calculator
a) Units consumed input.
b) Calculate bill:
a. First 100 units → ₹5/unit
b. Next 200 units → ₹7/unit
c. Above 300 → ₹10/unit
c) Add 5% surcharge if bill > ₹2000."""
units = int(input("Enter units consumed: "))
if units <= 100:
    bill = units * 5
elif units <= 300:
    bill = (100 * 5) + (units - 100) * 7
else:
    bill = (100 * 5) + (200 * 7) + (units - 300) * 10
if bill > 2000:
    bill = bill + (bill * 5 / 100)
print("Total Electricity Bill =", bill)
```

```python
"""
1. Automorphic Number
• Check whether square of number ends with same number.
• Example: 25**2 = 625 → Automorphic."""
n = int(input("Enter a number: "))
square = n * n
temp = n
count = 0
while temp > 0:
    count = count + 1
    temp = temp // 10
if square % (10 ** count) == n:
    print("Automorphic Number")
else:
    print("Not Automorphic")

```

```python
# 2. Palindrome Number Series: Print all palindrome numbers (same forward/backward) between given range.
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

for n in range(start, end + 1):
    temp = n
    rev = 0

    while temp > 0:
        digit = temp % 10
        rev = rev * 10 + digit
        temp = temp // 10

    if n == rev:
        print(n)
```

```python
# 3. LCM & HCF Using Loops (without built-in functions.)
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
# HCF
hcf = 1
for i in range(1, min(a, b) + 1):
    if a % i == 0 and b % i == 0:
        hcf = i
# LCM
lcm = (a * b) // hcf
print("HCF =", hcf)
print("LCM =", lcm)
```

```python
#4. Password Retry System: Allow max 3 attempts and Block after 3 failures.
password = "1234"
attempt = 0

while attempt < 3:

    user = input("Enter password: ")

    if user == password:
        print("Login Successful")
        break

    else:
        attempt = attempt + 1
        print("Wrong Password")

if attempt == 3:
    print("Account Blocked")
```

```python
# 5. Print different star patterns (triangle, pyramid, diamond).
# right triangle pattern
n = 5

for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print()
    
#pyramid pattern
n = 5

for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()

#diamond pattern
n = 5

# Upper part
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()

# Lower part
for i in range(n - 1, 0, -1):
    for j in range(n - i):
        print(" ", end="")

    for k in range(2 * i - 1):
        print("*", end="")

    print()

#
```

## String

```python
# 1. Palindrome String Check: Check whether a string is palindrome (ignore spaces & case).
s = input("Enter a string: ")

s = s.replace(" ", "").lower()

rev = s[::-1]

if s == rev:
    print("Palindrome String")
else:
    print("Not Palindrome")
```

```python
# 2. Find the longest word in a sentence.
s = input("Enter a sentence: ")

words = s.split()

longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest Word =", longest)
```

```python
#3. Check whether two strings are anagrams. Example: "listen" & "silent"
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

s1 = s1.lower()
s2 = s2.lower()

if sorted(s1) == sorted(s2):
    print("Anagram")
else:
    print("Not Anagram")
```

```python
#4. Remove duplicate characters from a string.
s = input("Enter a string: ")

result = ""

for ch in s:
    if ch not in result:
        result = result + ch

print("After removing duplicates:", result)
```

```python
#5. Compress string: "aaabbc" → "a3b2c1
s = input("Enter a string: ")

result = ""
count = 1

for i in range(len(s) - 1):

    if s[i] == s[i + 1]:
        count = count + 1

    else:
        result = result + s[i] + str(count)
        count = 1

result = result + s[-1] + str(count)

print("Compressed String:", result)
```
